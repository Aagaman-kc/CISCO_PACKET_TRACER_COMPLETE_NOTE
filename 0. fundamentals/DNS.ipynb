# <p style="font-size:32px; color:#1f6fb2; border-bottom: 1px solid #1f6fb2; padding-bottom:5px;">🌍 DNS (Domain Name System)</p>

## 🔹 What is DNS?
DNS is the **phonebook of the internet**.

Humans remember names like:
google.com

Computers understand only:
IP addresses like 142.250.183.14

👉 **DNS translates domain names into IP addresses.**

Without DNS, you would need to remember IPs for every website.

---

## 🔹 Why DNS Exists
- Humans cannot remember IP addresses
- IP addresses change frequently
- DNS provides abstraction and scalability
- Enables load balancing and redundancy

📌 DNS is **one of the most critical and trusted systems** on the internet.

---

## 🔹 DNS Analogy (Very Important)
Think of DNS like **asking directions**:

You: “Where is google.com?”  
DNS: “Go to 142.250.183.14”

You don’t care **how** DNS finds it — only that it works.

---

## 🔹 DNS Uses Which Protocol?
DNS mainly uses:
- **UDP port 53** → fast lookups
- **TCP port 53** → large responses (zone transfers, DNSSEC)

Why UDP by default?
- Faster
- Low overhead
- One request → one response

---

## 🔹 DNS Resolution Process (Step-by-Step)

### 1️⃣ User Query
You type:
google.com

Your computer asks:
“Do I already know this?”

---

### 2️⃣ Local Cache
If IP is cached → DONE  
If not → ask DNS resolver

---

### 3️⃣ Recursive Resolver (ISP / Router / Cloud)
This resolver does the hard work **on your behalf**.

---

### 4️⃣ Root DNS Server
Resolver asks:
“Who knows about .com?”

Root replies:
“Ask the .com TLD servers.”

---

### 5️⃣ TLD Server (.com)
Resolver asks:
“Who knows google.com?”

TLD replies:
“Ask Google’s authoritative server.”

---

### 6️⃣ Authoritative DNS Server
Final answer:
“google.com = 142.250.183.14”

Resolver returns this to you.

🎉 Connection starts.

---

## 🔹 DNS Record Types (VERY IMPORTANT)

| Record | Purpose |
|------|--------|
| **A** | Domain → IPv4 address |
| **AAAA** | Domain → IPv6 address |
| **CNAME** | Alias to another domain |
| **MX** | Mail server |
| **NS** | Authoritative name server |
| **TXT** | Verification, SPF, DKIM |
| **PTR** | Reverse DNS (IP → domain) |

---

## 🔹 Example
google.com → A → 142.250.183.14
mail.google.com → MX → mail server


---

## 🔹 DNS Caching
DNS responses are cached for a time called **TTL (Time To Live)**.

Benefits:
- Faster browsing
- Less DNS traffic
- Reduced server load

Risk:
- Poisoned cache = dangerous

# <p style="font-size:28px; color:#c0392b; border-bottom: 1px solid #c0392b; padding-bottom:5px;">🔥 How Attackers Abuse DNS</p>

## 🔹 1. DNS Spoofing / Poisoning
Attacker injects **fake DNS responses**.

Victim thinks:
google.com → real site

Actually goes to:
malicious server

Result:
- Credential theft
- Malware delivery
- Silent MITM

---

## 🔹 2. Rogue DNS Server
Attacker controls DNS server via:
- Rogue DHCP
- Manual DNS override
- Compromised router

Victim traffic is **fully redirected**.

---

## 🔹 3. DNS Hijacking
Attacker changes:
- Router DNS
- ISP DNS
- Domain registrar DNS records

Mass-scale attack:
- Thousands of victims
- Fake websites
- Crypto theft

---

## 🔹 4. DNS Tunneling
DNS used to:
- Bypass firewalls
- Exfiltrate data
- Maintain C2 channels

Because DNS is:
- Allowed everywhere
- Trusted by default

---

## 🔹 5. Subdomain Takeover
Unused DNS records point to cloud services.

Attacker claims the resource → controls subdomain.

Very common bug bounty issue.

---

# <p style="font-size:28px; color:#16a085; border-bottom: 1px solid #16a085; padding-bottom:5px;">🛡️ DNS Security & Defense</p>

## 🔹 DNSSEC
Digitally signs DNS records.

Prevents:
- DNS spoofing
- Cache poisoning

Downside:
- Complex
- Not universally deployed

---

## 🔹 Secure DNS Providers
Examples:
- Cloudflare (1.1.1.1)
- Google (8.8.8.8)
- Quad9 (9.9.9.9)

---

## 🔹 Monitoring & Detection
Watch for:
- Unexpected DNS servers
- High-volume DNS queries
- Random-looking domains
- Long subdomain strings

---

## 🔹 Network Segmentation
Limit DNS access between:
- Users
- Servers
- IoT
- Guests
