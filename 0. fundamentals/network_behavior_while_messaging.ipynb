<h2>1️⃣ Connecting to the Local Network (Wi-Fi / Ethernet)</h2>
<p>When your laptop/phone connects to a router:</p>
<ul>
<li>Wi-Fi authentication or Ethernet link is established.</li>
<li>The device joins the local network (LAN).</li>
<li>Now the device needs network configuration.</li>
</ul>
<p>Protocol used:</p>
<p><strong>Dynamic Host Configuration Protocol</strong></p>

<h3>DHCP Steps</h3>
<p><strong>DHCP Discover</strong><br>
Your device broadcasts: "Is there a DHCP server?"</p>

<p><strong>DHCP Offer</strong><br>
Router responds: "You can use IP: 192.168.1.20"</p>

<p><strong>DHCP Request</strong><br>
Your device asks to use that IP.</p>

<p><strong>DHCP Acknowledgement</strong><br>
Router confirms and sends configuration.</p>

<p>Your device receives:</p>
<ul>
<li>IP address: 192.168.1.20</li>
<li>Subnet mask: 255.255.255.0</li>
<li>Default gateway: 192.168.1.1</li>
<li>DNS server: (example 8.8.8.8)</li>
<li>Lease time</li>
</ul>
<p>Now your device can communicate on the network.</p>

<h2>2️⃣ You Want to Send a Message</h2>
<p>Suppose you send a message to: <strong>friend.com</strong></p>
<p>Computers cannot use domain names. They must convert the domain into an IP address.</p>

<h2>3️⃣ DNS Resolution</h2>
<p>Protocol used: <strong>Domain Name System</strong></p>
<p>Your computer asks: "What is the IP address of friend.com?"</p>

<h3>DNS Lookup Process</h3>
<ul>
<li>Check local DNS cache</li>
<li>Ask router DNS</li>
<li>Router forwards request to ISP resolver</li>
<li>Resolver queries:
<ul>
<li>Root DNS server</li>
<li>TLD server (.com)</li>
<li>Authoritative server</li>
</ul>
</li>
</ul>
<p>Finally DNS returns: <strong>friend.com → 104.26.10.78</strong></p>
<p>Now your computer knows the destination IP address.</p>

<h2>4️⃣ Determine Where to Send the Packet</h2>
<p>Your computer checks if the destination is in the same network.</p>
<p>Example:</p>
<ul>
<li>Your IP: 192.168.1.20</li>
<li>Destination: 104.26.10.78</li>
</ul>
<p>These are not in the same subnet. Therefore the packet must go to the default gateway (router).</p>

<h2>5️⃣ Finding Router MAC Address</h2>
<p>Computers send frames using MAC addresses on the LAN.</p>
<p>Protocol used: <strong>Address Resolution Protocol</strong></p>
<p>Your computer broadcasts: "Who has 192.168.1.1 ?"</p>
<p>Router replies: "192.168.1.1 is at MAC AA:BB:CC:DD:EE:FF"</p>
<p>Now your computer knows the router’s MAC address.</p>

<h2>6️⃣ Packet Creation (Encapsulation)</h2>
<p>Your message moves through the networking layers.</p>
<p>Layers:</p>
<ul>
<li>Application</li>
<li>Transport</li>
<li>Network</li>
<li>Data Link</li>
<li>Physical</li>
</ul>

<h3>Application Layer</h3>
<p>Example protocols:</p>
<ul>
<li>Hypertext Transfer Protocol</li>
<li>Hypertext Transfer Protocol Secure</li>
</ul>
<p>Your message is created.</p>

<h3>Transport Layer</h3>
<p>Protocols:</p>
<ul>
<li>Transmission Control Protocol</li>
<li>User Datagram Protocol</li>
</ul>
<p>For messaging or web communication usually TCP.</p>
<p>TCP adds:</p>
<ul>
<li>Source port</li>
<li>Destination port</li>
<li>Sequence numbers</li>
</ul>

<h3>Network Layer</h3>
<p>Protocol: Internet Protocol</p>
<p>Adds:</p>
<ul>
<li>Source IP: 192.168.1.20</li>
<li>Destination IP: 104.26.10.78</li>
</ul>

<h3>Data Link Layer</h3>
<p>Frame created with:</p>
<ul>
<li>Source MAC: your device</li>
<li>Destination MAC: router</li>
</ul>

<h3>Physical Layer</h3>
<p>Bits transmitted through Wi-Fi radio waves or Ethernet electrical signals.</p>

<h2>7️⃣ Router Receives Packet</h2>
<ul>
<li>Removes Ethernet header</li>
<li>Reads destination IP</li>
<li>Checks routing table</li>
</ul>
<p>Example:</p>
<ul>
<li>Destination: 104.26.10.78</li>
<li>Next hop: ISP gateway</li>
</ul>
<p>Router forwards packet.</p>

<h2>8️⃣ Network Address Translation</h2>
<p>Your device has a private IP. Example: 192.168.1.20</p>
<p>But the internet needs public IPs.</p>
<p>Technique used: <strong>Network Address Translation</strong></p>
<p>Router changes: 192.168.1.20 → 103.x.x.x</p>
<p>Router stores mapping: 103.x.x.x:45000 → 192.168.1.20:51500</p>
<p>This mapping is stored in the NAT table.</p>

<h2>9️⃣ ISP Network</h2>
<p>The packet enters your ISP infrastructure.</p>
<p>Routers inside ISP:</p>
<ul>
<li>Access router</li>
<li>Aggregation router</li>
<li>Core router</li>
</ul>
<p>Each router performs destination IP lookup and next-hop forwarding.</p>
<p>Internal routing protocols: Open Shortest Path First</p>

<h2>🔟 Global Internet Backbone</h2>
<p>The packet travels across international networks.</p>
<p>Large ISPs exchange routes using Border Gateway Protocol.</p>
<p>Possible path:</p>
<ul>
<li>Nepal ISP</li>
<li>India transit provider</li>
<li>Singapore Internet Exchange</li>
<li>US backbone provider</li>
<li>Destination hosting network</li>
</ul>
<p>Transmission occurs through fiber optic cables, Internet exchange points, and undersea cables.</p>

<h2>1️⃣1️⃣ Destination Data Center</h2>
<p>The packet arrives at the server network.</p>
<p>Typical path:</p>
<ul>
<li>Edge router</li>
<li>Load balancer</li>
<li>Application server</li>
</ul>
<p>If using HTTPS:</p>
<p>Protocol: Transport Layer Security</p>
<p>Server decrypts the packet.</p>

<h2>1️⃣2️⃣ Server Processes Message</h2>
<p>Server application performs:</p>
<ul>
<li>User authentication</li>
<li>Message validation</li>
<li>Recipient lookup</li>
</ul>
<p>Example database query: Find user "friend", Check active connection</p>
<p>Two cases occur.</p>

<h2>1️⃣3️⃣ If Friend Is Online</h2>
<p>Your friend's device already has a persistent connection to the server.</p>
<p>The server sends the message immediately through that connection.</p>

<h2>1️⃣4️⃣ If Friend Is Offline</h2>
<p>Server stores the message in database.</p>
<p>When friend opens the app: Server → deliver pending messages</p>

<h2>1️⃣5️⃣ Sending Message to Friend's Network</h2>
<p>Server sends packets to your friend.</p>
<p>Path: Server → US ISP → Friend home router → Friend device</p>
<p>Same routing process occurs.</p>

<h2>1️⃣6️⃣ Friend Router NAT</h2>
<p>Friend router uses NAT too.</p>
<p>Example mapping:</p>
<ul>
<li>Public IP: 66.x.x.x, Port: 49001 → Private IP: 192.168.0.12</li>
</ul>
<p>Router forwards packet to friend’s device.</p>

<h2>1️⃣7️⃣ Friend Device Receives Packet</h2>
<p>Reverse process (decapsulation): Physical → Data link → Network → Transport → Application</p>
<p>TCP reassembles packets. Application receives the message.</p>

<h2>1️⃣8️⃣ Message Displayed</h2>
<p>The app running on the phone processes the message. Notification appears: "New message from you". Your friend reads it.</p>

<h2>1️⃣ Your Friend Opens the App</h2>
<p>Suppose your friend opens a messaging app (WhatsApp, Discord, etc.).</p>
<p>The phone initiates an outgoing connection to the service server using Transmission Control Protocol and Transport Layer Security.</p>
<p>Example connection:</p>
<ul>
<li>Friend phone: 192.168.0.12:53000</li>
<li>Friend router (NAT)</li>
<li>Public IP: 66.120.14.5:45001</li>
<li>Messaging server: 157.240.x.x:443</li>
</ul>

<h2>2️⃣ NAT Mapping Is Created</h2>
<p>Friend’s router performs Network Address Translation</p>
<p>It creates a mapping table:</p>
<pre>
Private IP            Public Mapping
192.168.0.12:53000 → 66.120.14.5:45001
</pre>
<p>This table stays alive while the connection exists.</p>

<h2>3️⃣ Server Stores Connection Information</h2>
<p>Server records:</p>
<pre>
UserID: friend123
Connection IP: 66.120.14.5
Port: 45001
Socket: active TCP session
</pre>
<p>The server knows this TCP connection belongs to your friend. The server uses the public NAT address.</p>

<h2>4️⃣ Now You Send a Message</h2>
<p>Your device sends a message to the server. Server receives: Sender: you, Receiver: friend123, Message: "hello"</p>
<p>Server checks its connection table: friend123 → active connection, IP: 66.120.14.5, Port: 45001</p>

<h2>5️⃣ Server Sends Message Through Existing TCP Connection</h2>
<p>The server writes the message into that TCP session. Packet path: Server → Internet → 66.120.14.5:45001</p>

<h2>6️⃣ Friend Router Uses NAT Table</h2>
<p>The router sees destination 66.120.14.5:45001. NAT table: 66.120.14.5:45001 → 192.168.0.12:53000</p>
<p>Router forwards packet to 192.168.0.12 (friend’s phone).</p>

<h2>7️⃣ Friend Phone Receives the Packet</h2>
<p>The phone receives the packet through the TCP connection, decrypts TLS → app. The app displays: "New message from you".</p>

<h2>🔑 The Important Networking Idea</h2>
<ul>
<li>Servers do not discover your friend behind NAT.</li>
<li>Friend device initiates the connection first</li>
<li>NAT mapping is created</li>
<li>Server keeps the connection alive</li>
<li>Server sends messages through that connection</li>
</ul>
<p>This technique is called <strong>Persistent outbound connection</strong></p>
<p>Routers block incoming connections unless the device started the connection first. Messaging apps keep a long-lived connection alive using small heartbeat packets.</p>
