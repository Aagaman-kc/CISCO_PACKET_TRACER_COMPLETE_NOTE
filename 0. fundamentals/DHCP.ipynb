# <p style="font-size:32px; color:#8e44ad; border-bottom: 1px solid #8e44ad; padding-bottom:5px;">📡 DHCP (Dynamic Host Configuration Protocol)</p>

## 🔹 What is DHCP?
DHCP is a **network protocol that automatically assigns IP configuration** to devices when they join a network.

Without DHCP, every device would need **manual configuration**, which is slow, error-prone, and impossible at scale.

DHCP automatically provides:
- IP address
- Subnet mask
- Default gateway
- DNS server(s)

📌 **Real-world analogy:**  
DHCP is like a **hotel front desk**:
- You enter the hotel (network)
- Front desk gives you a room number (IP)
- Tells you which hallways to use (gateway)
- Tells you who to call for services (DNS)

---

## 🔹 Why DHCP Exists
- Eliminates manual IP configuration
- Prevents IP conflicts
- Enables plug-and-play networking
- Essential for enterprise, cloud, and Wi-Fi networks

Without DHCP, **modern networks would collapse**.

---

## 🔹 DHCP Uses UDP (Not TCP)
DHCP uses:
- **UDP port 67** (server)
- **UDP port 68** (client)

Why UDP?
- Client has **no IP yet**
- No TCP handshake possible
- Broadcasts are required

---

## 🔹 DHCP DORA Process (VERY IMPORTANT)

DHCP works using **DORA**:

### 1️⃣ Discover
Client broadcasts:
“I need an IP, is there any DHCP server?”

Source IP: `0.0.0.0`  
Destination IP: `255.255.255.255`

---

### 2️⃣ Offer
DHCP server responds:
“I can give you this IP.”

Includes:
- Offered IP
- Subnet mask
- Gateway
- Lease time

---

### 3️⃣ Request
Client replies:
“I want that IP from you.”

This confirms which server was chosen.

---

### 4️⃣ Acknowledge
Server finalizes:
“IP is yours for X time.”

Client is now fully connected.

---

## 🔹 DHCP Lease
A **lease** is a temporary IP assignment.

Lease lifecycle:
- 50% → renewal attempt
- 87.5% → rebinding attempt
- Expired → IP revoked

This allows IP reuse and scalability.

---

## 🔹 Where DHCP Is Used
- Home routers
- Enterprise LANs
- Wi-Fi networks
- Cloud VPCs (AWS, Azure, GCP)
- Containers and Kubernetes
- VPNs

# <p style="font-size:28px; color:#c0392b; border-bottom: 1px solid #c0392b; padding-bottom:5px;">🔥 How Attackers Abuse DHCP</p>

## 🔹 1. Rogue DHCP Server (Most Dangerous)
Attacker sets up a **fake DHCP server**.

Victims receive:
- Attacker-controlled gateway
- Attacker-controlled DNS

Result:
- Man-in-the-Middle (MITM)
- Credential theft
- Traffic interception
- Malware injection

📌 This attack works **without malware**.

---

## 🔹 2. DHCP Starvation Attack
Attacker floods DHCP server with fake requests.

Effects:
- DHCP pool exhausted
- Legit users cannot get IPs
- Network denial-of-service

Often used **before** launching a rogue DHCP server.

---

## 🔹 3. DNS Hijacking via DHCP
Attacker supplies malicious DNS server.

Victim consequences:
- Fake websites
- Credential harvesting
- Malware downloads
- Silent redirection

User sees **correct URL**, but wrong IP.

---

## 🔹 4. Network Mapping
DHCP reveals:
- IP ranges
- Subnet size
- Gateway IP
- DNS servers

Attackers use this for:
- Reconnaissance
- Lateral movement planning
- Target prioritization

---

# <p style="font-size:28px; color:#16a085; border-bottom: 1px solid #16a085; padding-bottom:5px;">🛡️ Defense & Security Controls</p>

## 🔹 DHCP Snooping
Switch feature that:
- Trusts DHCP responses only from legit ports
- Blocks rogue DHCP servers

Essential in enterprise networks.

---

## 🔹 Port Security
Limits number of MAC addresses per port.

Stops:
- Starvation attacks
- MAC flooding

---

## 🔹 Static IP for Critical Systems
Servers, routers, firewalls should **not use DHCP**.

---

## 🔹 Network Segmentation
Separate:
- User devices
- Servers
- IoT
- Guests

Limits DHCP-based attacks.

What is a "Port" exactly?

Think of an IP address like the street address of an apartment building. If you send a letter just to the building, the mailman knows which house it goes to, but not which person inside should get it.

A Port is like the apartment number. It tells the phone’s operating system exactly which "room" (or application) the data belongs to so that your web browser doesn't accidentally get the data meant for your Spotify stream.
Why do DNS and DHCP have specific numbers?

In the networking world, these are called "Well-Known Ports" (0–1023). They are standardized so that every device on earth knows where to look for basic services.
Service	Port(s)	Why it uses these
DNS	53	To translate "google.com" into an IP address. It uses UDP for speed and TCP for large data transfers.
DHCP	67 & 68	67 is for the Server (the router) to listen for requests. 68 is for the Client (your phone) to receive its assigned IP.
Do modern phones have these ports?

Yes, but with a slight catch: you don't see them because they are "logical," not physical.

    Physical Port: The USB-C or Lightning hole where you plug in your charger.

    Logical Port: An invisible software "mailbox" inside your phone's memory.

When you walk into a coffee shop and connect to their Wi-Fi, your phone immediately opens Port 68 and shouts, "Is there a DHCP server here? I need an IP address!" The router hears this on Port 67 and sends the info back to your phone’s Port 68. Once that’s done, your phone uses Port 53 every time you type a website name into your browser.

    Note: Modern phones also use "Dynamic Ports" (numbers like 49152–65535). These are temporary ports your phone picks at random to handle things like a specific Instagram post loading or a single WhatsApp message.
