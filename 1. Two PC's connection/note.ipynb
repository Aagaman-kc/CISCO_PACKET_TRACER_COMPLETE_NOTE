# Two pc connection

***We have two PC's connected with the copper cross over***

***We can assign the ip address by double tapping pc -> desktop -> ip configuration -> changing the IPv4 address***

### 192.168.1.1
***Here, 192.168.1.1 is a class-c ip address***<br>
***so, 192.168.1 is a network part and last 1 is part of host***

***NOTE*** <br>
***Network part must be simpler between two PC's for communication, else there is no connection established***
***192.168.1.1 and 192.168.1.2 have same network part but have different host***

| FIRST OCTET RANGE | CLASS | NETWORK/HOST BITS | DEFAULT MASK |
| :--- | :---: | :---: | :--- |
| 1–126 | A | 8/24 | 255.0.0.0 |
| 128–191 | B | 16/16 | 255.255.0.0 |
| 192–223 | C | 24/8 | 255.255.255.0 |
| 224–239 | D | Multicast | N/A |
| 240–255 | E | Experimental | N/A |

### To see the connection, we ping one PC from terminal of other
***for this double tap pc -> desktop -> command prompt -> (ping 192.168.1.2)***
