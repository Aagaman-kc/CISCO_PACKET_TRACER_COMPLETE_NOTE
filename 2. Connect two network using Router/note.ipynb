# Network Fundamentals: Router vs Switch

## 📡 Router
- **Function**: Connects different networks, routes data between them
- **Analogy**: Post office directing mail between neighborhoods
- **Key Features**:
  - Assigns IP addresses (DHCP)
  - Network Address Translation (NAT)
  - Firewall protection
  - Often includes wireless access point

## 🔄 Switch
- **Function**: Connects devices within same network
- **Analogy**: Telephone operator connecting calls in same building
- **Key Features**:
  - Directs data to specific devices using MAC addresses
  - Operates within single network
  - Faster than hubs (sends to specific device only)


## 🔗 Key Points
1. **Different Networks**: 192.168.1.x ≠ 192.168.2.x
2. **Same Network Communication**: Switch handles (192.168.1.2 → 192.168.1.3)
3. **Different Network Communication**: Router required (192.168.1.2 → 192.168.2.2)
4. **Modern Routers Include**: Router + Switch + Wireless AP

## 📶 Wireless Functionality
- Modern routers have built-in wireless access points
- Creates Wi-Fi network (SSID)
- Wireless devices get IP addresses like wired devices
- Still connects to internal switch portion

## 🎯 Quick Rules
- **Router**: Between networks, uses IP addresses
- **Switch**: Within network, uses MAC addresses  
- **Gateway**: Router interface (e.g., 192.168.1.1)
- **Subnet Mask**: 255.255.255.0 (/24) defines network boundaries

***As we can see we have set the unique ip address for each pc and laptop based on network but we also have set the default gateway***
## 🎯 What is Default Gateway?
- **Definition**: The IP address of the router interface in your network
- **Purpose**: The "exit door" for traffic leaving your local network
- **Example**: If your computer is 192.168.1.2, gateway might be 192.168.1.1

## 🔍 Why It's Essential

### 1. **Internet Access**
- All internet traffic goes through gateway
- Gateway connects local network to external networks

### 2. **Communication with Other Networks**

### 3. **Routing Decisions**

**Without front door**: Can't leave house
**With front door**: Can go anywhere
