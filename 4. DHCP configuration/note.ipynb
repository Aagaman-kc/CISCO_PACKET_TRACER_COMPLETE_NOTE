# DHCP
**DHCP allow the router to automatically assign ip address to computer or devices**

**We have assigned the DHCP to both pc and laptop and no need to manually type the ip address, subnet mask and default gateway**

**Here, we setup the router ip address only**

**As we can see, the g0/0/0 is up with assigned ip address such that we lastly used wr command tp write the changes**

# now our task is to configure dhcp

**Here, we have configured dhcp mode**

**see, we got the ip address automatically**

```python
# Phase 1: Basic Interface Configuration
en                   # Enter Privileged EXEC mode (like admin)
config t             # Enter Global Configuration mode
hostname R1          # Set router name to R1

# Configure GigabitEthernet 0/0/0 interface
int g0/0/0           # Enter interface config mode
ip add 192.168.1.1 255.255.255.0  # Assign IP and subnet mask
no sh                # Enable the interface (bring it up)

# Phase 2: Verification and Saving
sh ip int br         # Show IP interface brief (verify Status: up, Protocol: up)
wr                   # Save configuration to NVRAM

# Phase 3: DHCP Pool Configuration
ip dhcp excluded-address 192.168.1.1   # Exclude router IP from DHCP pool
ip dhcp pool aagaman                   # Create DHCP pool named 'aagaman'
network 192.168.1.0 255.255.255.0     # Define DHCP address range
default-router 192.168.1.1            # Set default gateway for clients
^Z                                     # Exit config mode
```
