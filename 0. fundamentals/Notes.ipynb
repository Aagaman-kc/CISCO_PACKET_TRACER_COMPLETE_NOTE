# <p style="font-size:32px; color:#289a0c; border-bottom: 1px solid #289a0c; padding-bottom:5px;">🌐 Subnetting</p>

### <span style="color:#2980B9;">🔹 What is a Subnet?</span>
A **subnet** is a logical partition of a larger IP network. It helps organize devices, reduce broadcast traffic, and significantly improve security through isolation.

---

### <span style="color:#2980B9;">🔹 Key Benefits</span>
* **Performance:** Minimizes network congestion.
* **Security:** Enables **segmentation** to block lateral movement.
* **Organization:** Group devices by department (e.g., HR, Engineering).
* **Cloud:** Essential for AWS VPC and Azure VNet configurations.

---

### <span style="color:#2980B9;">🔹 IP Address Structure</span>
IPv4 addresses consist of **32 bits** divided into four **8-bit octets**.



**The Subnet Mask:** Determines which part of the address is the **Network** (street) and which is the **Host** (house).
* **Binary 1:** Network bit.
* **Binary 0:** Host bit.

| Type | Example |
| :--- | :--- |
| **IP Address** | `192.168.1.10` |
| **Subnet Mask** | `255.255.255.0` |

---

### <span style="color:#2980B9;">🔹 CIDR & Common Ranges</span>
**CIDR** (Classless Inter-Domain Routing) uses a slash `/` to indicate the number of network bits.

| CIDR | Subnet Mask | Usable Hosts | Use Case |
| :--- | :--- | :--- | :--- |
| **/24** | `255.255.255.0` | **254** | Standard LAN |
| **/26** | `255.255.255.192` | **62** | Smaller Segments |
| **/27** | `255.255.255.224` | **30** | Server Clusters |
| **/30** | `255.255.255.252` | **2** | Point-to-Point Links |



---

### <span style="color:#2980B9;">🔹 Example Breakdown (`/24`)</span>
For the network **`192.168.1.0/24`**:
* **Network ID:** `192.168.1.0`
* **Usable IPs:** `192.168.1.1` – `192.168.1.254`
* **Broadcast:** `192.168.1.255`

---

### <span style="color:#2980B9;">🔹 Security Perspective</span>
Subnetting is a primary tool for **Network Segmentation**. By placing sensitive data in its own subnet, you can use firewalls to restrict access, preventing an attacker from hopping between systems.

### <span style="color:#2980B9;">🔹 Example</span>
* **10.0.0.0/16 means:**
Network Part (Fixed): 10.0.x.x <br>
Host Part (Variable): x.x.0.1 through x.x.255.254<br>
Network ID: 10.0.0.0<br>
Subnet Mask: 255.255.0.0<br>
Total IP Addresses: 65,536<br>
