# <p style="font-size:32px; color:#8e44ad; border-bottom: 1px solid #8e44ad; padding-bottom:5px;">📡 DHCP (Dynamic Host Configuration Protocol)</p>

## 🔹 What is DHCP?
DHCP is a **network protocol that automatically assigns IP configuration** to devices when they join a network.

Without DHCP, every device would need **manual configuration**, which is slow, error-prone, and impossible at scale.

DHCP automatically provides:
- IP address
- Subnet mask
- Default gateway
- DNS server(s)

📌 **Real-world analogy:**  
DHCP is like a **hotel front desk**:
- You enter the hotel (network)
- Front desk gives you a room number (IP)
- Tells you which hallways to use (gateway)
- Tells you who to call for services (DNS)

---

## 🔹 Why DHCP Exists
- Eliminates manual IP configuration
- Prevents IP conflicts
- Enables plug-and-play networking
- Essential for enterprise, cloud, and Wi-Fi networks

Without DHCP, **modern networks would collapse**.

---

## 🔹 DHCP Uses UDP (Not TCP)
DHCP uses:
- **UDP port 67** (server)
- **UDP port 68** (client)

Why UDP?
- Client has **no IP yet**
- No TCP handshake possible
- Broadcasts are required

---

## 🔹 DHCP DORA Process (VERY IMPORTANT)

DHCP works using **DORA**:

### 1️⃣ Discover
Client broadcasts:
“I need an IP, is there any DHCP server?”

Source IP: `0.0.0.0`  
Destination IP: `255.255.255.255`

---

### 2️⃣ Offer
DHCP server responds:
“I can give you this IP.”

Includes:
- Offered IP
- Subnet mask
- Gateway
- Lease time

---

### 3️⃣ Request
Client replies:
“I want that IP from you.”

This confirms which server was chosen.

---

### 4️⃣ Acknowledge
Server finalizes:
“IP is yours for X time.”

Client is now fully connected.

---

## 🔹 DHCP Lease
A **lease** is a temporary IP assignment.

Lease lifecycle:
- 50% → renewal attempt
- 87.5% → rebinding attempt
- Expired → IP revoked

This allows IP reuse and scalability.

---

## 🔹 Where DHCP Is Used
- Home routers
- Enterprise LANs
- Wi-Fi networks
- Cloud VPCs (AWS, Azure, GCP)
- Containers and Kubernetes
- VPNs

# <p style="font-size:28px; color:#c0392b; border-bottom: 1px solid #c0392b; padding-bottom:5px;">🔥 How Attackers Abuse DHCP</p>

## 🔹 1. Rogue DHCP Server (Most Dangerous)
Attacker sets up a **fake DHCP server**.

Victims receive:
- Attacker-controlled gateway
- Attacker-controlled DNS

Result:
- Man-in-the-Middle (MITM)
- Credential theft
- Traffic interception
- Malware injection

📌 This attack works **without malware**.

---

## 🔹 2. DHCP Starvation Attack
Attacker floods DHCP server with fake requests.

Effects:
- DHCP pool exhausted
- Legit users cannot get IPs
- Network denial-of-service

Often used **before** launching a rogue DHCP server.

---

## 🔹 3. DNS Hijacking via DHCP
Attacker supplies malicious DNS server.

Victim consequences:
- Fake websites
- Credential harvesting
- Malware downloads
- Silent redirection

User sees **correct URL**, but wrong IP.

---

## 🔹 4. Network Mapping
DHCP reveals:
- IP ranges
- Subnet size
- Gateway IP
- DNS servers

Attackers use this for:
- Reconnaissance
- Lateral movement planning
- Target prioritization

# <p style="font-size:28px; color:#16a085; border-bottom: 1px solid #16a085; padding-bottom:5px;">🛡️ Defense & Security Controls</p>

## 🔹 DHCP Snooping
Switch feature that:
- Trusts DHCP responses only from legit ports
- Blocks rogue DHCP servers

Essential in enterprise networks.

---

## 🔹 Port Security
Limits number of MAC addresses per port.

Stops:
- Starvation attacks
- MAC flooding

---

## 🔹 Static IP for Critical Systems
Servers, routers, firewalls should **not use DHCP**.

---

## 🔹 Network Segmentation
Separate:
- User devices
- Servers
- IoT
- Guests

Limits DHCP-based attacks.

# <p style="font-size:28px; color:#2980b9; border-bottom: 1px solid #2980b9; padding-bottom:5px;">📋 DHCP Cheat Sheet</p>

## 🔹 Ports
- UDP 67 → DHCP Server
- UDP 68 → DHCP Client

---

## 🔹 DORA Flow
Discover → Offer → Request → Acknowledge

---

## 🔹 Common Attacks
- Rogue DHCP
- DHCP Starvation
- DNS Hijacking
- MITM

---

## 🔹 Blue Team Detection
- Multiple DHCP offers
- Unexpected gateway changes
- DNS server changes
- High DHCP request rate

---

## 🔹 Tools (Learning / Lab)
- `tcpdump -i eth0 port 67 or 68`
- `tshark -i eth0 -f "udp port 67 or 68"`
- Wireshark DHCP filter: `bootp`

---

## 🔹 Red Team (Academic Only)
- DHCP starvation simulation
- Rogue DHCP lab environments
- Traffic interception demos

⚠️ Only in **isolated labs** or **CTFs**

---

## 🔹 Final Takeaway
DHCP is:
- Invisible
- Critical
- Highly trusted

Which makes it:
- Extremely powerful
- Extremely dangerous when abused

Understanding DHCP deeply is **mandatory for cybersecurity**.
