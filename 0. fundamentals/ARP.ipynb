# <p style="font-size:32px; color:#8e44ad; border-bottom: 1px solid #8e44ad; padding-bottom:5px;">🔁 ARP (Address Resolution Protocol)</p>

## 🔹 What is ARP?
ARP is used to **map an IP address to a MAC address** inside a local network.

Computers communicate using:
- **IP address** → logical identity
- **MAC address** → physical identity

👉 ARP answers the question:
“Who has this IP address?”

---

## 🔹 Why ARP is Needed
When your system wants to send data to:
192.168.1.10

It must know:
**Which MAC address belongs to that IP**

ARP makes this possible.

Without ARP:
- IP packets cannot be delivered on LAN
- Ethernet communication fails

---

## 🔹 ARP Works Only Inside a LAN
ARP is:
- **Layer 2.5** protocol (between L2 & L3)
- Not routable
- Never leaves your local network

Routers do **not forward ARP packets**.

---

## 🔹 ARP Analogy
Imagine a classroom:

You shout:
“Who is Roll No. 10?”

Person replies:
“I am Roll No. 10!”

Now you know **who to talk to**.

---

## 🔹 ARP Packet Types

| Type | Description |
|----|-----------|
| **ARP Request** | Broadcast: “Who has IP X?” |
| **ARP Reply** | Unicast: “IP X is at MAC Y” |

---

## 🔹 ARP Resolution Process (Step-by-Step)

### 1️⃣ Check ARP Cache
Your system checks:
“Do I already know this MAC?”

If yes → DONE

---

### 2️⃣ ARP Request (Broadcast)
If not cached, system sends:
Who has 192.168.1.10?
Tell 192.168.1.5


Sent to:
FF:FF:FF:FF:FF:FF

Everyone hears it.

---

### 3️⃣ ARP Reply (Unicast)
The owner replies:
192.168.1.10 is at AA:BB:CC:DD:EE:FF


Only to requester.

---

### 4️⃣ Cache Storage
The mapping is stored temporarily in **ARP table**.

---

## 🔹 ARP Cache
ARP entries have a timeout.

Benefits:
- Faster communication
- Less broadcast traffic

Risk:
- Can be poisoned

---

# <p style="font-size:28px; color:#c0392b; border-bottom: 1px solid #c0392b; padding-bottom:5px;">☠️ ARP Poisoning / ARP Spoofing</p>

## 🔹 What is ARP Poisoning?
ARP has **no authentication**.

Any device can say:
“I am the owner of this IP”

👉 Attacker sends **fake ARP replies** to victims.

---

## 🔹 Why ARP Is Vulnerable
- Stateless protocol
- No verification
- Trust-based
- Accepts unsolicited replies

This makes ARP extremely dangerous.

---

## 🔹 How ARP Poisoning Works

### 1️⃣ Attacker Sends Fake ARP Replies
To victim:
Gateway IP → Attacker MAC

To gateway:
Victim IP → Attacker MAC

---

### 2️⃣ Man-in-the-Middle (MITM)
Attacker now sits between:
Victim ↔ Router

Traffic passes through attacker.

---

### 3️⃣ What Attacker Can Do
- Sniff passwords
- Hijack sessions
- Modify packets
- Inject malware
- Downgrade HTTPS

---

## 🔹 Types of ARP Attacks

| Attack | Description |
|------|------------|
| **MITM** | Silent traffic interception |
| **DoS** | Drop packets |
| **Session Hijacking** | Steal cookies |
| **Traffic Redirection** | Fake websites |

---

## 🔹 Common Tools Used by Attackers
- arpspoof
- ettercap

(Used in labs & CTFs like pwn.college)

# <p style="font-size:28px; color:#16a085; border-bottom: 1px solid #16a085; padding-bottom:5px;">🛡️ ARP Defense & Mitigation</p>

## 🔹 Static ARP Entries
Manually bind IP → MAC.

Pros:
- Prevents spoofing

Cons:
- Not scalable

---

## 🔹 Dynamic ARP Inspection (DAI)
Switch-level protection.

Checks ARP packets against DHCP bindings.

Very effective.

---

## 🔹 Port Security
Limits MAC addresses per switch port.

Prevents rogue devices.

---

## 🔹 Encryption
Even if ARP is poisoned:
- HTTPS
- SSH
- TLS

Protect data contents.

---

## 🔹 Network Segmentation
Separate:
- Users
- Servers
- Guests
- IoT

Reduces attack surface.

---

# <p style="font-size:28px; color:#2980b9; border-bottom: 1px solid #2980b9; padding-bottom:5px;">📋 ARP Cheat Sheet</p>

## 🔹 Important Facts
- ARP is Layer 2 based
- Uses broadcast
- Works only in LAN
- No authentication

---

## 🔹 Wireshark Filters
arp


---

## 🔹 Red Flags (Detection)
- Multiple ARP replies for same IP
- Gateway MAC changing frequently
- Unsolicited ARP replies
- High ARP traffic

---

## 🔹 Blue Team Monitoring
- Enable ARP inspection
- Log ARP changes
- Alert on MAC-IP mismatch

---

## 🔹 Final Takeaway
ARP is:
- Simple
- Essential
- Insecure by design

ARP poisoning is one of the **easiest and most powerful LAN attacks**.

If you understand ARP deeply, you understand **how MITM attacks really work**.
