# 📡 Wireless Configuration Using an Access Point (Connected to a Switch)

## 1. Introduction
Wireless networking allows devices to connect to a network **without physical cables** by using **radio signals (Wi-Fi)**.

In many networks, wireless devices connect through an **Access Point (AP)**.

Basic communication flow:

```
Wireless Device → Access Point → Switch → Network / Internet
```

The **Access Point connects wireless devices to the wired LAN** through a switch.

---

# 2. What is an Access Point (AP)?

An **Access Point (AP)** is a networking device that **provides Wi-Fi connectivity and connects wireless devices to a wired network**.

It acts as a **bridge between wireless devices and the wired LAN**.

### Example Wireless Devices
- Laptop
- Smartphone
- Tablet
- Smart TV
- IoT devices

These devices connect **wirelessly to the AP**, and the AP forwards data to the **switch and the rest of the network**.

---

# 3. Why Do We Use an Access Point?

Without an Access Point:
- Devices must connect using **Ethernet cables**
- **Mobility is limited**

With an Access Point:
- Devices can **move freely**
- Multiple devices can **connect wirelessly**

### Common Places Where APs Are Used
- Homes
- Offices
- Universities
- Airports
- Cafes

---

# 4. Basic Network Topology

```
        Laptop / Phone
             |
           Wi-Fi
             |
        Access Point
             |
        Ethernet Cable
             |
           Switch
             |
      Router / Server / Internet
```

### Data Flow
1. Wireless device connects to the **Access Point**
2. The **Access Point sends data to the switch**
3. The **switch forwards data to other devices or the router**

---

# 5. Components in This Setup

## Access Point (AP)
- Provides **Wi-Fi signal**
- Connects **wireless devices to the wired LAN**
- Configurable settings include:
  - SSID
  - Security
  - Channel

---

## Switch
A **switch connects multiple wired devices in a LAN**.

Examples of devices connected to a switch:
- Computers
- Servers
- Routers
- Access Points

The switch forwards frames using **MAC addresses**.

---

## Wireless Clients
Devices that connect to Wi-Fi.

Examples:
- Laptop
- Mobile phone
- Tablet
- IoT devices

---

# 6. Important Wireless Concepts

## SSID (Service Set Identifier)

The **SSID is the name of the wireless network**.

Users select the SSID when connecting to Wi-Fi.

---

## Wireless Security

Security protects the wireless network from **unauthorized access**.

| Security | Description |
|--------|-------------|
| Open | No password |
| WPA | Older wireless security |
| WPA2 | Most commonly used |
| WPA3 | New and stronger security |

Example configuration:

```
SSID: OfficeWiFi
Password: strongpassword123
Security: WPA2
```

---

## Wireless Channels

Wi-Fi uses **radio frequency channels** to transmit data.

Common frequency bands:

- **2.4 GHz**
- **5 GHz**

Channels help **reduce interference between wireless networks**.

---

# 7. Advantages of Access Points

- Enables **wireless connectivity**
- Provides **mobility**
- Allows **multiple devices to connect**
- Reduces need for **network cables**

---

# 8. Limitations

- Wireless interference
- Potentially lower speed than wired networks
- Security risks if **not properly configured**

---

# 📌 Key Concept

```
Wireless Devices → Access Point → Switch → Network
```

The **Access Point provides Wi-Fi**, while the **switch connects wired network devices**.
