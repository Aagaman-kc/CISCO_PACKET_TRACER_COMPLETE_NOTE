<h1>Routing (Beginner Friendly)</h1>

<p>
<b>Routing</b> is the process of sending data from one network to another.
Whenever you open a website, your data does not go directly to the destination.
Instead, it passes through <b>multiple routers</b> on the way.
</p>

<p>
A <b>router</b> decides which direction the data should move next so that it reaches
the correct destination.
</p>

<hr>

<h2>How a Router Makes a Decision</h2>

<p>When a router receives a data packet, it checks:</p>

<ul>
  <li><b>Destination IP address</b> – Where does the packet want to go?</li>
  <li><b>Routing table</b> – A list of known paths to other networks</li>
  <li><b>Best path</b> – The most suitable way to forward the packet</li>
</ul>

<hr>

<h2>Types of Routing</h2>

<h3>1️⃣ Static Routing</h3>

<p>
<b>Static routing</b> means the network administrator <b>manually tells the router</b>
where to send packets.
</p>

<p>
Think of it like giving someone <b>fixed directions</b>:
“Always take this road to reach the office.”
</p>

<ul>
  <li>Routes are set by hand</li>
  <li>They do not change automatically</li>
  <li>Best for small or simple networks</li>
</ul>

<h4>What is “Next Hop”?</h4>

<p>
The <b>next hop</b> is the <b>very next router</b> that a packet should be sent to.
</p>

<p>
<b>Analogy:</b><br>
If you want to reach a friend’s house, the <b>next hop</b> is the first bus or road you take —
not the final destination.
</p>

<p><b>Example:</b></p>

<pre>
Destination Network: 192.168.2.0/24
Next Hop: 192.168.1.1
</pre>

<p>
This means:<br>
“To reach the 192.168.2.0 network, send the packet first to the router at 192.168.1.1.”
</p>

<p>
✔️ Simple and reliable<br>
❌ If the route breaks, the admin must fix it manually
</p>

<hr>

<h3>2️⃣ Routing Information Protocol (RIP)</h3>

<p>
<b>RIP</b> is a <b>dynamic routing protocol</b>.
Routers using RIP automatically share information with each other.
</p>

<p>
Instead of fixed directions, routers <b>talk to each other</b> and learn routes on their own.
</p>

<h4>What is a “Hop”?</h4>

<p>
A <b>hop</b> means <b>one router</b> that a packet passes through.
</p>

<p>
<b>Hop count</b> = number of routers between the source and the destination.
</p>

<p>
<b>Analogy:</b><br>
If you change buses to reach a place:
<ul>
  <li>1 bus change = 1 hop</li>
  <li>3 bus changes = 3 hops</li>
</ul>
</p>

<h4>How RIP Works</h4>

<ul>
  <li>Chooses the path with the <b>fewest hops</b></li>
  <li>Maximum hop count allowed is <b>15</b></li>
  <li>Automatically updates routes when the network changes</li>
</ul>

<p><b>Example:</b></p>

<ul>
  <li>Path A: 2 routers (2 hops)</li>
  <li>Path B: 4 routers (4 hops)</li>
</ul>

<p>
RIP selects <b>Path A</b> because it has fewer hops.
</p>

<p>
✔️ Easy to use<br>
❌ Not suitable for large networks
</p>

<hr>

<h2>Static Routing vs RIP (Beginner View)</h2>

<table border="1" cellpadding="8" cellspacing="0">
  <tr>
    <th>Feature</th>
    <th>Static Routing</th>
    <th>RIP</th>
  </tr>
  <tr>
    <td>Who decides the route?</td>
    <td>Administrator</td>
    <td>Routers automatically</td>
  </tr>
  <tr>
    <td>Route changes</td>
    <td>No (manual)</td>
    <td>Yes (automatic)</td>
  </tr>
  <tr>
    <td>Metric used</td>
    <td>None</td>
    <td>Hop count</td>
  </tr>
  <tr>
    <td>Best for</td>
    <td>Small networks</td>
    <td>Small to medium networks</td>
  </tr>
</table>

<hr>

<h2>Key Takeaway</h2>

<ul>
  <li>
    <b>Static Routing</b> → Fixed paths set by humans
  </li>
  <li>
    <b>RIP</b> → Routers learn paths automatically using hop count
  </li>
</ul>

<h1>🖧 Static Routing Configuration (Two Routers)</h1>

<h2>📌 Network Topology</h2>
<ul>
  <li><b>R1 LAN:</b> 192.168.1.0/24</li>
  <li><b>R2 LAN:</b> 192.168.2.0/24</li>
  <li><b>Link between routers:</b> 10.1.1.0/8</li>
</ul>

<hr>

<h2>🔹 Router 1 Configuration (R1)</h2>

<pre>
enable
configure terminal
hostname R1

interface gigabitEthernet0/0/1
ip address 192.168.1.1 255.255.255.0
no shutdown

interface gigabitEthernet0/0/0
ip address 10.1.1.1 255.0.0.0
no shutdown

end
</pre>

<div class="success">
✅ Verify interface status:<br>
<pre>show ip interface brief</pre>
💾 Save configuration:<br>
<pre>write</pre>
</div>

<h3>🔹 Static Route on Router 1</h3>

<pre>
configure terminal
ip route 192.168.2.0 255.255.255.0 10.1.1.2
end
</pre>

<div class="note">
🧠 <b>Explanation:</b><br>
192.168.2.0 → destination network<br>
255.255.255.0 → subnet mask<br>
10.1.1.2 → next hop (Router R2’s IP)<br>
“To reach 192.168.2.0 network, send packets to R2.”
</div>

<hr>

<h2>🔹 Router 2 Configuration (R2)</h2>

<pre>
enable
configure terminal
hostname R2

interface gigabitEthernet0/0/1
ip address 192.168.2.1 255.255.255.0
no shutdown

interface gigabitEthernet0/0/0
ip address 10.1.1.2 255.0.0.0
no shutdown

end
</pre>

<div class="success">
💾 Save configuration:<br>
<pre>write</pre>
</div>

<h3>🔹 Static Route on Router 2</h3>

<pre>
configure terminal
ip route 192.168.1.0 255.255.255.0 10.1.1.1
end
</pre>

<div class="note">
🧠 <b>Explanation:</b><br>
“To reach 192.168.1.0 network, forward packets to Router R1.”
</div>

<hr>

<h2>✅ Key Takeaways</h2>
<ul>
  <li><b>Network ID</b> ends with .0</li>
  <li><b>Next hop</b> = IP of the next router</li>
  <li>Static routing = manually telling routers where to send traffic</li>
  <li>Interfaces must be <b>UP / UP</b> for routing to work</li>
</ul>
